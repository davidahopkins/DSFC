## Creating a function in python

To create a function in python we use *def* followed by the name for our function, then in () we add an parameters that the function will use, we place a colon : after the (). the following lines are the instruction for the function and should be intendented to be included in the function. See example below.  

```python
def wall_surface(width, length, height):
    perimeter = (width * 2) + (length * 2)
    wall_surf = perimeter * height
    return(wall_surf)
```

```python
wall_surface(10, 20, 9)
```
